```python
import jax.numpy as jnp
import numpy as np
import matplotlib.pyplot as plt

def generate_test_datasets(num_datasets=4, lattice_size=10, num_timepoints=100):
    datasets = []
    rng = np.random.default_rng(42)

    for i in range(num_datasets):
        alpha = round(rng.uniform(0.0, 2.0), 2)
        beta = round(rng.uniform(0.0, 1.0), 2)
        
        # Simulate increasing binding over time, with noise
        frac_bound_series = jnp.clip(jnp.linspace(0, 1, num_timepoints) + 0.1 * jnp.sin(jnp.linspace(0, 10, num_timepoints) + i), 0, 1)

        # Create binary lattice snapshots based on frac_bound
        snapshots = jnp.array([
            jnp.array(rng.random((lattice_size, lattice_size)) < fb, dtype=int)
            for fb in frac_bound_series
        ])

        # Simulate non-uniform time points
        times = jnp.cumsum(jnp.abs(jnp.diff(jnp.concatenate([jnp.array([0.0], rng.uniform(0.05, 0.2, num_timepoints)]))))

        datasets.append((snapshots, times, alpha, beta))

    return datasets

def plot_multiple_fraction_bound(datasets, filename='fraction_bound_vs_time.png'):
    """
    Plot fraction of bound receptors over time for multiple (alpha, beta) values and save figure.
    
    Parameters:
    - datasets: list of tuples, each tuple contains:
        (snapshots, times, alpha, beta)
    - filename: name of the file to save the plot (default 'fraction_bound_vs_time.png')
    """

    plt.figure(figsize=(10, 5))

    for i, (snapshots, times, alpha, beta) in enumerate(datasets):
        # Type check
        if not hasattr(snapshots, 'reshape'):
            raise TypeError(f"Dataset index {i} has invalid 'snapshots': expected array, got {type(snapshots)}")

        # Compute fraction bound for each timepoint
        frac_bound = jnp.mean(snapshots.reshape(len(times), -1), axis=1)
        label = f"α={alpha}, β={beta}"
        plt.plot(times, frac_bound, lw=2, label=label)

    plt.xlabel("Time (s)")
    plt.ylabel("Fraction Bound")
    plt.title("Fraction of Bound Receptors Over Time")
    plt.ylim(0, 1)
    plt.grid(True)
    plt.legend(title="Coupling Parameters")
    plt.tight_layout()
    plt.savefig(filename)
    plt.close()

test_data = generate_test_datasets()
plot_multiple_fraction_bound(test_data)
```
